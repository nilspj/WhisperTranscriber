## 4 Set Input and Output File Names

The cell below contains two lines:

    input_filename = ""
    output_filename = ""

* In input_filename, enter the exact name of your uploaded audio file, including the file extension (e.g. .mp3, .wav, .mp4).
* In output_filename, enter the name of the output text file, ending with .txt.

Example:

    input_filename = "audio_file.mp4"
    output_filename = "audio_file_transcription.txt"

After editing the file names, click the Play ▶️ button in the top-left corner of the cell before continuing.

```python
input_filename = "input.mp4"
output_filename = "output.txt"
```

## 5 Start Transcription

Click the Play ▶️ button in the cell below to begin transcription.

A spinning circle will appear while the transcription is running.
This can take 1–30 minutes, depending on audio length and model size.

Once finished:
*   The output .txt file will appear in the Files panel on the left.
*   Hover over the file, click the three dots, and select Download to save your transcription.

```python
import whisper
import numpy as np
import torch
torch.cuda.empty_cache()
model = whisper.load_model("large")
result = model.transcribe(input_filename)
with open(output_filename, "w", encoding="utf-8") as txt:
    txt.write(result["text"])
torch.cuda.empty_cache()
```

## Having Problems?
Occasionally, things may go wrong. Try the following:
1. Restart the transcription process from Step 1.
2. Click Runtime → Restart session, confirm, then repeat Step 1.
3. Double-check that the file name and extension are correct. The file may be in an unsupported format — try converting it and start again.
4. The audio file may be too long. Try splitting it into smaller parts (up to ~1 hour per file). Repeat Step 1 for each part.
